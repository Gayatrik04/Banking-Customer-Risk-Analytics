```python
import pandas as pd

customers = pd.read_csv("../data/clean/customers_clean.csv")
accounts = pd.read_csv("../data/clean/accounts_clean.csv")
loans = pd.read_csv("../data/clean/loans_clean.csv")
transactions = pd.read_csv("../data/clean/transactions_clean.csv")
branches = pd.read_csv("../data/clean/branches_clean.csv")
```

```python
validation_summary = pd.DataFrame({
    "Dataset": [ "customers","accounts","loans", "transactions","branches"],
    "Missing Values": [
        customers.isnull().sum().sum(),
        accounts.isnull().sum().sum(),
        loans.isnull().sum().sum(),
        transactions.isnull().sum().sum(),
        branches.isnull().sum().sum()
    ],
    "Duplicate Rows": [
        customers.duplicated().sum(),
        accounts.duplicated().sum(),
        loans.duplicated().sum(),
        transactions.duplicated().sum(),
        branches.duplicated().sum()
    ]
})

validation_summary
```

```python
def validate_dataset(df, name, primary_key):
    print("=" * 60)
    print(f"{name.upper()} DATASET")
    print("=" * 60)

    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    print("\nMissing Values")
    print(df.isnull().sum())

    print("\nDuplicate Rows")
    print(df.duplicated().sum())

    print("\nDuplicate Primary Keys")
    print(df[primary_key].duplicated().sum())

    print("\nData Types")
    print(df.dtypes)

    print("\n")
    
validate_dataset(customers, "Customers", "customer_id")
validate_dataset(accounts, "Accounts", "account_id")
validate_dataset(loans, "Loans", "loan_id")
validate_dataset(transactions, "Transactions", "transaction_id")
validate_dataset(branches, "Branches", "branch_id")
```

```python
invalid_customers = accounts[
    ~accounts["customer_id"].isin(customers["customer_id"])
]

print("Invalid Customer IDs:", len(invalid_customers))
```

```python
invalid_branches = accounts[
    ~accounts["branch_id"].isin(branches["branch_id"])
]

print("Invalid Branch IDs:", len(invalid_branches))
```

```python
invalid_loans = loans[
    ~loans["customer_id"].isin(customers["customer_id"])
]

print("Invalid Loan Customers:", len(invalid_loans))
```

```python
invalid_transactions = transactions[
    ~transactions["customer_id"].isin(customers["customer_id"])
]

print("Invalid Transaction Customers:", len(invalid_transactions))
```

```python
invalid_accounts = transactions[
    ~transactions["account_id"].isin(accounts["account_id"])
]

print("Invalid Transaction Accounts:", len(invalid_accounts))
```
