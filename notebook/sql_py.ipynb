```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine(
    "mysql+pymysql://root:Gayatri%401311@localhost:3306/banking_customer_risk_analytics"
)

customers = pd.read_csv("../data/clean/customers_clean.csv")
branches = pd.read_csv("../data/clean/branches_clean.csv")
accounts = pd.read_csv("../data/clean/accounts_clean.csv")
loans = pd.read_csv("../data/clean/loans_clean.csv")
transactions = pd.read_csv("../data/clean/transactions_clean.csv")
```

```python
customers.to_sql("customers",con=engine,if_exists="append",index=False,chunksize=5000,method="multi")
```

```python
branches.to_sql("branches",con=engine,if_exists="append",index=False,chunksize=5000,method="multi")
```

```python
accounts.to_sql("accounts",con=engine,if_exists="append",index=False,chunksize=5000,method="multi")
```

```python
loans.to_sql("loans",con=engine,if_exists="append",index=False,chunksize=5000,
method="multi")
```

```python
transactions.to_sql("transactions",con=engine,if_exists="append",index=False,chunksize=10000,method="multi")
```

```python
print(pd.read_sql("SELECT COUNT(*) AS total FROM customers", engine))
print(pd.read_sql("SELECT COUNT(*) AS total FROM branches", engine))
print(pd.read_sql("SELECT COUNT(*) AS total FROM accounts", engine))
print(pd.read_sql("SELECT COUNT(*) AS total FROM loans", engine))
print(pd.read_sql("SELECT COUNT(*) AS total FROM transactions", engine))
```
